```python
import numpy as np
from DBSCAN_contoller import DBSCAN

# Dataset fourni (Étudiants/Points en 2D)
X_train = np.array([
    [1, 2], [1.5, 1.8], [2, 2], [8, 7], [8, 8], [8.5, 7.5],
    [25, 80], [1, 1], [0.5, 1], [50, 50]
])

# Paramètres de base (assez stricts)
model = DBSCAN(eps=2.5, min_samples=3)
labels = model.fit(X_train)

print("--- RÉSULTATS SESSION NORMALE ---")
for i in range(len(X_train)):
    status = "ADMIS (Cluster " + str(labels[i]) + ")" if labels[i] >= 0 else "ÉCHEC (Bruit/Isolé)"
    print(f"Point {X_train[i]} -> {status}")
```
