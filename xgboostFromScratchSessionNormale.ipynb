```python
import numpy as np
from XGBoostClassifier import XGBoostClassifier

# 1. Données d'entraînement (Historique)
X_train = np.array([[7.5, 90], [1.0, 10], [5.0, 60], [2.5, 40], [8.0, 95], [3.0, 30]])
y_train = np.array([18, 4, 12, 7, 19, 8])

# 2. Entraînement du modèle
model = XGBoostClassifier(n_estimators=20, learning_rate=0.1)
model.fit(X_train, y_train)

# 3. Session Normale pour nouveaux étudiants
X_session_normale = np.array([[4.5, 55], [2.0, 25], [8.5, 98]])
notes_normale = model.predict_note(X_session_normale)
verdict_normale = model.get_status(notes_normale)

print("--- RÉSULTATS SESSION NORMALE ---")
for i, note in enumerate(notes_normale):
    print(f"Étudiant {i+1}: Note {note:.2f}/20 -> {verdict_normale[i]}")
```
