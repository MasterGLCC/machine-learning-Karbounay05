```python
import numpy as np
from DBSCAN_contoller import DBSCAN

X_train = np.array([
    [1, 2], [1.5, 1.8], [2, 2], [8, 7], [8, 8], [8.5, 7.5], 
    [25, 80], [1, 1], [0.5, 1], [50, 50] 
])

# 1. On lance d'abord la session normale
model_normal = DBSCAN(eps=2.5, min_samples=3)
labels_n = model_normal.fit(X_train)

print("--- ANALYSE DES ÉCHECS & RATTRAPAGE ---")

for i in range(len(X_train)):
    if labels_n[i] != -1:
        print(f"Point {X_train[i]} : Déjà classé (Cluster {labels_n[i]})")
    else:
        print(f"Point {X_train[i]} : ÉCHEC -> Tentative de Rattrapage...")
        
        # Le "rattrapage" consiste à être plus tolérant sur la distance
        # On augmente le rayon (Epsilon) pour ce point précis
        model_ratrap = DBSCAN(eps=15.0, min_samples=2) 
        labels_r = model_ratrap.fit(X_train)
        
        if labels_r[i] != -1:
            print(f"   => RÉUSSI au rattrapage ! (Rejoint le Cluster {labels_r[i]})")
        else:
            print(f"   => ÉCHEC DÉFINITIF (Trop éloigné de tout)")
```
