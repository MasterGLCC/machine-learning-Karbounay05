```python
import numpy as np

# Importation de votre classe personnalisée
from XGBoostClassifier import XGBoostClassifier

# =================================================================
# 1. PRÉPARATION & ENTRAÎNEMENT (Nécessaire pour initialiser 'model')
# =================================================================
# Données historiques pour que le modèle sache comment noter
X_train = np.array([
    [7.5, 90], [1.0, 10], [5.0, 60], [2.5, 40],
    [8.0, 95], [3.0, 30], [6.0, 80], [0.5, 5]
])
y_train = np.array([18, 4, 12, 7, 19, 8, 15, 2])

# Initialisation et entraînement
model = XGBoostClassifier(n_estimators=15, learning_rate=0.2, max_depth=2)
model.fit(X_train, y_train)

# =================================================================
# 2. SIMULATION SESSION NORMALE
# =================================================================
# On teste sur des étudiants qui ont des notes proches de la limite
X_test = np.array([
    [3.5, 40],  # Étudiant A : Risque d'échec
    [5.5, 65],  # Étudiant B : Proche de la moyenne
    [2.0, 20]   # Étudiant C : Grand risque d'échec
])

notes_n = model.predict_note(X_test)

# =================================================================
# 3. LOGIQUE DE RATTRAPAGE (DEUXIÈME CHANCE)
# =================================================================
print("\n" + "="*50)
print("SYSTÈME DE GESTION DES EXAMENS (XGBOOST)")
print("="*50)

for i in range(len(X_test)):
    print(f"\n--- Étudiant {i+1} ---")
    print(f"Stats Initiales : {X_test[i][0]}h étude, {X_test[i][1]}% participation")
    print(f"Note Session Normale : {notes_n[i]:.2f}/20")

    if notes_n[i] >= 10:
        print("Résultat : ADMIS (Session Normale)")
    else:
        print("Résultat : ÉCHEC -> Admission en Session de Rattrapage")

        # --- PHASE DE RATTRAPAGE ---
        # On simule que l'étudiant double ses heures d'étude pour le rattrapage
        X_rattrapage = X_test[i].copy()
        X_rattrapage[0] = X_rattrapage[0] * 1.5  # Augmente le temps d'étude de 50%
        X_rattrapage[1] = min(100, X_rattrapage[1] + 15) # +15% de présence/révision

        # Nouvelle prédiction basée sur les nouveaux efforts
        note_r = model.predict_note(np.array([X_rattrapage]))[0]

        print(f"Nouvelles Stats   : {X_rattrapage[0]:.1f}h étude, {X_rattrapage[1]}% participation")
        print(f"Note Rattrapage   : {note_r:.2f}/20")

        if note_r >= 10:
            print("Verdict Final    : RÉUSSITE (Rattrapage)")
        else:
            print("Verdict Final    : ÉCHEC DÉFINITIF")

print("\n" + "="*50)
```
